```python
class Employee:
  ''' This is the employee class tosafe the data of employees'''
  starttime = 9
  headoffice = "newyork"
  def __init__(self , name , role , salary):
    self.name = name
    self.role = role
    self.salary = salary
  def displayInfo(self):
    print(f"{self.name} is {self.role} and his salary is {self.salary}")
emp1 = Employee("Musa" , "Engineer" , "150,000")
emp2 = Employee("Ali" , "Automobile" , "200,000")


emp1.displayInfo()
```
