```python
class Vehicle:
  run = True
```

```python
class Car(Vehicle):
  def makesound(self):
    print("Gheen")
class Bike(Vehicle):
  def makesound(self):
    print("Ghaaan")
# As we all know dog ha sno relationship with vehicle but As per the condition
# only a sinle func can it part of that particular parent class.
class Dog(Vehicle):
  def makesound(self):
    print("Whoof")
vehicles = [Car(),Bike(),Dog()]
for vehicl in vehicles:
  vehicl.makesound()
```

```python
class Animal:
  walk = True
class Dog(Animal):
  def Legs(self):
    print("Dog has 4 legs")
class Cat(Animal):
  def Legs(self):
    print("Cat has four leg")
class Humman:
  def Legs(self):
    print("Human has 2 legs")
organisms = [Dog(),Cat(),Humman()]
for organism in organisms:
  organism.Legs()
```
