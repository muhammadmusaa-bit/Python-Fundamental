```python
# while loop
# while something:
  # some condition
```

```python
i = 1 # initilization
while i <= 5:
  print("hello")
  i = i + 1
```

```python
c = 1
while c <= 10:
    print("Python")
    c = c + 1
```

```python
r = 1
while r <= 100:
  print("Zero")
  r = r + 1
```

```python
# Nusted loop
i = 1
while i <= 5:
  print("Hallo")
  i = i + 1
  s = 1
  while s <= 10:
    print("Musa")
    s = s + 1
```

```python
# for loop - sequences like list , range , string etc
```

```python
for i in range(1,10):
  print(i)
```

```python
for i in range(1,10):
  if i == 5:
    continue;
  else:
    print(i)
```

```python
contries = ["Pakistan" , "USA" , "UK" , "Germany"]
for i in contries:
  print(i)
```

```python
word = "Pakistan"
for i in word:
  print(i)
```

```python
# Table of 7
num = 7
for i in range(1,11):
  print(f"{num} * {i} = {num * i}")
```

```python
# Table of 9
num = 9
for i in range(1,11):
  print(f"{num} * {i} = {num * i}")
```

```python
# Table printer system
num = int(input("Enter the number"))
for i in range(1,11):
  print(f"{num} * {i} = {num * i}")
```

```python
# Factorial of 5
num = int(input("Enter the number"))
factorial = 1
for i in range(1 , num+1):
  factorial = factorial * num
print(factorial)
```
