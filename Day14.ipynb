```python
class Car:
  def __init__(self):
    self.model = 'Corolla'
    self.brand = "TOYOTA"
    self.color = "Black"
    self.power_type = "Fuel"
car1 = Car()
```

```python
print(f"Model of the car is {car1.model}")
print(f"Brand of the car is {car1.brand}")
print(f"Color of the car is {car1.color}")
print(f"Power type of the car is {car1.power_type}")
```
