```python
# Variable Length argument
def add(x , *y):
  z = x
  for i in y:
    z = z + i
  print(f"Total amount is {z}")
add(23,56,65,67,54,56,78,54,66)
```

```python
# Keyword Arrgument
def student(name , ** arrgument):
  print(f"Name: {name}")
  for key , value in arrgument.items():
    print(f"{key} : {value}")
```

```python
student(name ="Musa" , city = "Salkot" , classe = 9 , grade = 'A')
```
