```python
alist = [1,2,3,4,5]
print(list(map(lambda p: p*p , alist)))
```

```python
alist = [2,3,4,5,7]
print(list(map(lambda p:p+p , alist)))
```

```python
# Filter
```

```python
even = list(filter(lambda p:p%2 == 0 , alist))
even
```

```python
odd = list(filter(lambda p:p%2 != 0 , alist))
odd
```

```python
# Reduce
from functools import reduce
```

```python
add = reduce(lambda x,y:x+y , alist)
add
```

```python
sub = reduce(lambda x,y:x-y , alist)
sub
```

```python
product = reduce(lambda x,y:x*y , alist)
product
```
