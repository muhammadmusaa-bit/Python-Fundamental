```python
# Local variable
def hello():
  a = "hello"
  b = "World!"
  c = a + " " + b
  return c
hello()
```

```python
# Global Variable
def message(mess):
  print("I am Muhammad Musa")
  print("I learn in class 11")
mess = "I am 16 year old"
message(mess)

```

```python
# Lambda / Ananoumys
addition = lambda x:x+6
result = addition(45)
print(result)
```

```python
subtraction = lambda x,y:x-y
results = subtraction(12,4)
print(results)
```

```python
multiply = lambda x,y:x*y
result1 = multiply(12,2)
print(result1)
```
