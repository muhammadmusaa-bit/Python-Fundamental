```python
class Employee:
  def __init__(self , name , role , salary):
    self.name = name
    self.role = role
    self.salary = salary
  def displayinfo(self):
    print(f"{self.name} is {self.role} and has salary of {self.salary}")
  class Adress:
    def __init__(self , street , city):
      self.street = street
      self.city = city
    def adressInfo(self):
      print(f"Adress of the employee is :{self.street}-{self.city}")
Emp1 = Employee("Musa" , "Doctor" , "130k")
Emp2 = Employee("Ali" , "Engineer" , "150k")
emp1adress = Employee.Adress("123 main street" , "Berlin")
emp2adress = Employee.Adress("123 King street" , "Munich")
print(Emp1.displayinfo())
```

```python
print(Emp2.displayinfo())
```

```python
print(emp1adress.adressInfo())
```
