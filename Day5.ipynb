```python
score = int(input("Enter the score:- "))
if score == 100:
  print("You did amazing.....")
elif score >= 70:
  print("Congratulation! You passed")
elif (score >= 50) and (score <= 70):
  print("You failed but you can improve..")
else:
  print("You are failed badly")
print("Thank you for using our system")
```

```python
# Calculator
num1 = int(input("Enter the first number"))
num2 = int(input("Enter the seacond number"))
operator = input("Enter the operator:- ")
if operator == "+":
  print(num1 + num2)
elif operator == "-":
  print(num1 - num2)
elif operator == "*":
  print(num1 * num2)
elif operator == "**":
  print(num1 ** num2)
elif operator == "/":
  print(num1 / num2)
elif operator == "//":
  print(num1 // num2)
else:
  print("Invalid operator")
```
