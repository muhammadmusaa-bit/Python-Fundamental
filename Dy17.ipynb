```python
class Employee:
  headoffice = 'NewYork'
  def __init__(self , name , role , salary):
    self.name = name
    self.role = role
    self.salary = salary
    # Instance method
  def employeeInfo(self):
    print(f"{self.name} is an {self.role} and has salary of {self.salary}")
    # Class method
  @classmethod
  def headofficeInfo(cls):
    print(f"Headoffice is in {cls.headoffice}")
    # Static method
  @staticmethod
  def genericInfo():
    print("This is the information of employees")
emp1 = Employee("Musa" , "Ai Engineer" , "130k")
emp2 = Employee("Ali" , "Engineer" , "150k")
print(emp1.employeeInfo())
```

```python
print(emp2.employeeInfo())
```

```python
print(Employee.headofficeInfo())
```

```python
print(Employee.genericInfo())
```
