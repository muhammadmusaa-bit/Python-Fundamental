```python
# Inheritance
```

```python
class A:
  def method1(self):
    print("This is the code of method A1")
  def method2(self):
    print("This is the code of method A2")
class B(A):
  def method3(self):
    print("This is the code of method B1")
  def method4(self):
    print("This is the cde of method B2")
class C(B):
  def method5(self):
    print("This is code of method C1")
  def method6(self):
    print("This is the code of method C2")
```

```python
b = B()
```

```python
b.method3()
```

```python
b.method2()
```

```python
c = C()
```

```python
c.method6()
```

```python
c.method6()
```

```python
# Multi inheritance
```

```python
class A:
  def method1(self):
    print("This is the code of method A1")
  def method2(self):
    print("This is the code of method A2")
class B:
  def method3(self):
    print("This is the code of method B1")
  def method4(self):
    print("This is the cde of method B2")
class C(A,B):
  def method5(self):
    print("This is code of method C1")
  def method6(self):
    print("This is the code of method C2")
```

```python
c = C()
```

```python
c.method2()
```

```python
# Method calling in Inheritance
class A:
  def __init__(self):
    print("This is the construtor of A")
  def methodA1(self):
    print("This is the code of method A1")
  def methodA2(self):
    print("This is the ocde of method A2")
class B:
  def __init__(self):
    super.__init__()
    print("This is the constructor of B")
  def methodB3(self):
    print("This is the code of method B3")
  def methodB4(self):
    print("This is the code of method B4")
class C(A,B):
  def __init__(self):
    print("This is the constructor of C")
  def methodC5(self):
    print("This is the code of method C5")
  def methodC6(self):
    print("This is the code of method C6")
```

```python
c = C()
c.methodB4()
```

```python
c.methodC5()
```

```python
c.methodA1()
```
